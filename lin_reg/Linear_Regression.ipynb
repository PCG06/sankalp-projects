### Understanding Train/Test Split in Machine Learning

 Imagine training a student using only past exam papers, then testing them on the exact same papers. They'd score perfectly! But on a new exam, they might fail because they memorized answers rather than learning concepts.


Similarly, if we train a machine learning model on data and test it on the same data, we can't tell if the model learned real patterns or just memorized the examples. This is called overfitting.

By dividing data into distinct training and testing sections, we ensure an honest evaluation of how well the model performs on completely new, unseen data.

1. **Training Dataset**
This is the data your algorithm learns from. The model studies patterns, relationships, and features in this dataset to adjust its internal parameters.
Purpose: Teaching the model

2. **Test Dataset**
This is all new data for the model.  You use it to see how well your model adapts to new scenarios.
Purpose: Final performance evaluation

3. **Validation Dataset**
During training, you use this separate set to monitor how well the model is performing on unseen data. It helps you catch overfitting early and tune parameters without using the final test set.
Purpose: Monitoring during training and hyperparameter tuning

**Common Split Ratios**

The exact percentages depend on your situation and available data:

80% train / 20% test: Standard choice for most projects

70% train / 30% test: Used when you have abundant data

75% train / 25% test: Common middle ground

60% train / 20% validation / 20% test: When fine-tuning parameters

### Gain knowledge about least squares

**Least Squares** method is a statistical procedure to find the best fit for a set of data points by minimizing the sum of the offsets or residuals of points from the plotted curve.

**Calculate Line Of Best Fit**

A more accurate way of finding the line of best fit is the least square method.

Use the following steps to find the equation of line of best fit for a set of ordered pairs $(x_1,y_1),(x_2,y_2),...(x_n,y_n)$.

**Step 1:** Calculate the slope ‘m’ by using the following formula:

$$m = \frac{\sum \left ( x-\bar{x} \right )* \left ( y-\bar{y} \right )}{\sum \left ( x-\bar{x} \right )^{2}}$$


**Step 2:** Compute the y -intercept of the line by using the formula:

$$c = y - mx$$

**Step 3:** Substitute the values in the final equation

$$y = mx + c$$

* y: dependent variable
* m: the slope of the line
* x: independent variable
* c: y-intercept


As an example, we will try to find the least squares regression line for the below data set:

\begin{array} {|r|r|}\hline Hours Spent & Grade \\\hline 6 & 82 \\ \hline 10 & 88 \\ \hline 2 & 56 \\ \hline 4 & 64 \\ \hline 6 & 77 \\ \hline 7 & 92 \\ \hline 0 & 23 \\ \hline 1 & 41 \\ \hline 8 & 80 \\ \hline 5 & 59 \\ \hline 3 & 47 \\ \hline  \end{array}

#### Polynomial Features

characteristics that are produced by raising preexisting characteristics to an exponent are known as polynomial features.

A polynomial feature, for instance, would be the creation of a new feature (column) with values determined by squaring the values in $X$, such as $X^2$, if a dataset had a single input feature $X$. Each input variable in the dataset can be modified by repeating this method. Additionally, the degree-2 polynomial features of a two-dimensional input sample of the type $[a, b]$ are $[1, a, b, a^2, ab, b^2]$.

Polynomial features are therefore a form of feature engineering, which is the process of developing new input features based on preexisting features.The polynomial's "degree" determines how many features are added; for example, a degree of three adds two new variables for every input variable. Usually, a small degree, like two or three, is employed.

```python
def PolynomialRegression(degree=2, **kwargs):
    return make_pipeline(PolynomialFeatures(degree), LinearRegression(**kwargs))   # using a pipeline to string these operations together
```

```python
def make_data(N, err=1.0, rseed=1):
    # randomly sample the data
    rng = np.random.RandomState(rseed)
    X = rng.rand(N, 1) ** 2
    y = 10 - 1. / (X.ravel() + 0.1)
    if err > 0:
        y += err * rng.randn(N)
    return X, y

X, y = make_data(40)
```

```python
# learning curve for generated dataset with a second-order polynomial model and a ninth-order polynomial

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)

for i, degree in enumerate([2, 7]):
    N, train_lc, val_lc = learning_curve(PolynomialRegression(degree),
                                         X, y, cv=7,
                                         train_sizes=np.linspace(0.3, 1, 25))

    ax[i].plot(N, np.mean(train_lc, 1), color='blue', label='training score')
    ax[i].plot(N, np.mean(val_lc, 1), color='red', label='validation score')
    ax[i].hlines(np.mean([train_lc[-1], val_lc[-1]]), N[0], N[-1],
                 color='gray', linestyle='dashed')

    ax[i].set_ylim(0, 1)
    ax[i].set_xlim(N[0], N[-1])
    ax[i].set_xlabel('training size')
    ax[i].set_ylabel('score')
    ax[i].set_title('degree = {0}'.format(degree), size=14)
    ax[i].legend(loc='best')
```

### Regularized Linear Models

Regularizing the model, or constraining it, is a smart technique to lessen overfitting because it will be more difficult for the model to overfit the data if it has fewer degrees of freedom.

Regularization for a linear model is usually accomplished by limiting the model's weights.

There are three methods for limiting the weights:

1. Ridge Regression

2. Regression Lasso

3. The Elastic Net

#### **Ridge regression**

**Ridge regression** or **Tikhonov regularization** is the regularization technique that performs L2 regularization. It modifies the loss function by adding the penalty (shrinkage quantity) equivalent to the square of the magnitude of coefficients.

- **sklearn.linear_model.Ridge** is the module used to solve a regression model where the loss function is the linear least squares function and regularization is L2.

#### Ridge Regression with Scikit-Learn

```python
n_samples, n_features = 15, 10
rng = np.random.RandomState(0)
y = rng.randn(n_samples)
X = rng.randn(n_samples, n_features)

rdg = linear_model.Ridge(alpha = 0.5)                  # instantiate Ridge regressor
rdg.fit(X, y)
rdg.score(X,y)
```

#### **LASSO (Least Absolute Shrinkage and Selection Operator)**

LASSO is the regularisation technique that performs L1 regularisation. It modifies the loss function by adding the penalty (shrinkage quantity) equivalent to the summation of the absolute value of coefficients.

 - **sklearn.linear_model.Lasso** is a linear model, with an added regularisation term, used to estimate sparse coefficients.


#### Lasso Regression with Scikit-Learn

```python
# uses coordinate descent as the algorithm to fit the coefficients

Lreg = linear_model.Lasso(alpha = 0.5)
Lreg.fit([[0,0], [1, 1], [2, 2]], [0, 1, 2])
```

The model can forecast new values in the following ways after it has been fitted:

#### Elastic-Net Regression

A regularized regression technique called **Elastic-Net** linearly integrates the L1 and L2 penalties of the Lasso and Ridge regression techniques.
